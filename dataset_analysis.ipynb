```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data/train.csv')
```

```python
# How many missing values?
df.isnull().sum()
```

```python
missing_percent = df.isnull().mean() * 100
missing_percent
```

Strategy
Age: missing 20 % => replace with median
Cabin: missing 77% => Drop column
Embarked: missing 2% and categorical => replace with most frequent value (mode)

```python
df = df.drop(columns=["Cabin"])
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
df.isnull().sum()
```
