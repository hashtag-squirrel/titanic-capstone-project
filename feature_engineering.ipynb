```python
def compute_percentile_ranges(column):
    #Define percentile ranges
    ranges = [(0, 50), (50, 100), (100, 200), (200, 300), (300, 400), (400, 500), (500, 600)]

    percentiles = {(low, high): (column.quantile(low/600), column.quantile(high/600)) for low, high in ranges}

    counts = {}

    for r, (low, high) in percentiles.items():
        counts[r] = ((column >= low) & (column < high)).sum()
    
    return counts


print(compute_percentile_ranges(train['Fare']))
```
